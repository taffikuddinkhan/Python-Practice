```python
import numpy as np

# counting range of the given dataset
#manual steps to calculate the range of a give n data

array1 = np.array([10,20,30,40,50])
range_val = np.max(array1) - np.min(array1)
print("RANGE : " , range_val)

array2 = np.array([10,30,50,70,100])
range_val = np.max(array2) - np.min(array2)
print("RANGE : " , range_val)

# calculate the range using predefined function-------------------------------------------------------------------------------------------------------------------------------------------
print("Range of array 1: ",np.ptp(array1))
print("Range of array 2: ",np.ptp(array2))


```

```python
# calculating the variance of the data 
import numpy as np
val = np.array([10,20,30])
print("variance is : - ",np.var(val))
print("Median : - ",np.mean(val))
print("standard deviation : -",np.std(val))
```

```python
import matplotlib.pyplot as plt

student_data = [10,20,30,40,30,20,40,30,60]
plt.xlabel("MARKS")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.hist(student_data)  # histogram is to see data distribution
plt.show()
```
